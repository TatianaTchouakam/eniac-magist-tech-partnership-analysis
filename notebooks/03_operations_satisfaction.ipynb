## 03 — Operations & Satisfaction (Magist)

### Goal
Evaluate whether Magist’s operational performance and customer satisfaction meet ENIAC’s expectations for a potential tech partnership.

---

### Key Business Questions
- Is Magist operationally reliable (delivery performance)?
- Are customers generally satisfied with their experience?
- Could operations be a limiting factor for ENIAC?

---

### Key Metrics
- On-time delivery rate
- Average customer review score
- Share of positive reviews (4–5 stars)
- Review score distribution

---

### Data & Methodology
All operational and satisfaction KPIs are computed using **SQL (MySQL 8)**.

📁 **SQL reference files (GitHub):**
- `sql/01_business_questions_mysql.sql`  
  → Review classification and satisfaction logic  
- `sql/02_kpis_for_slides_mysql.sql`  
  → Final operational KPIs used in slides  

These KPIs are directly extracted from:
- `orders`
- `order_reviews`
- `order_items`

---

## Results — Operations & Satisfaction

### 1️⃣ Delivery Performance
- **On-time delivery rate:** **91.9%**

The vast majority of orders are delivered within the expected timeframe, indicating **strong logistics performance**.

---

### 2️⃣ Customer Satisfaction
- **Average review score:** **4.08 / 5**
- **Positive reviews (4–5 stars):** **76.8%**

#### Review distribution:
- ⭐⭐⭐⭐⭐ (5): 56,593  
- ⭐⭐⭐⭐ (4): 18,929  
- ⭐⭐⭐ (3): 8,120  
- ⭐⭐ (2): 3,158  
- ⭐ (1): 11,571  

Customer feedback is **largely positive**, with a strong concentration of high ratings.

---

## Business Interpretation
Magist demonstrates:
- ✅ Reliable delivery performance
- ✅ High customer satisfaction
- ✅ Operational maturity at scale

Operations and service quality are **not the limiting factors** for a potential partnership.

---

## Conclusion (Slide-ready)
Magist is a **reliable and operationally strong platform**.  
However, operational excellence alone **cannot compensate for the limited scale and pricing of the tech segment**.

➡️ The main constraint remains **strategic alignment**, not execution.
