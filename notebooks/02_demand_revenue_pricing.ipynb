## 02 — Demand, Revenue & Pricing (Magist)

### Goal
Assess whether customer demand, revenue generation, and price positioning on Magist are aligned with ENIAC’s premium tech strategy.

---

### Key Questions
- Is customer demand for tech products significant on Magist?
- How does tech revenue compare to non-tech revenue?
- Is Magist’s average tech price aligned with ENIAC’s premium pricing?

---

### Key Metrics
- Tech vs Non-Tech orders (demand proxy)
- Tech vs Non-Tech revenue
- Average tech item price on Magist
- ENIAC average product price benchmark (€540)

---

### Expected Insight
While Magist shows steady overall growth, the tech segment may not generate sufficient demand or revenue to support ENIAC’s premium positioning.

## Results — Demand, Revenue & Pricing

### Demand (Orders)
Analysis of orders containing tech products shows that **tech-related orders represent a minority of total orders on Magist**.  
Customer demand on the platform is therefore **primarily driven by non-tech products**, indicating that tech is not the dominant purchase category.

---

### Revenue
Revenue analysis confirms that **tech products contribute a significantly smaller share of total revenue** compared to non-tech products.  
This suggests that even when tech items are purchased, their overall impact on revenue remains limited.

---

### Price Positioning
The **average tech item price on Magist is approximately €120**.  
In contrast, **ENIAC’s average product price is €540**, clearly positioning ENIAC in the premium segment.

This highlights a **strong price gap** between Magist’s tech market and ENIAC’s pricing strategy.

---

## Business Insight
There is a clear mismatch between Magist’s tech demand and revenue levels and ENIAC’s premium price positioning.  
While Magist performs well as a generalist marketplace, its tech segment currently lacks both the demand scale and pricing power required to support ENIAC’s premium tech expansion strategy.
