## 02 — Demand, Revenue & Pricing (Magist)

### Goal
Assess whether customer demand, revenue generation, and price positioning on Magist are aligned with **ENIAC’s premium tech strategy**.

---

### Key Business Questions
- Is customer demand for tech products significant?
- How does tech revenue compare to non-tech revenue?
- Is Magist’s average tech price aligned with ENIAC’s premium pricing?

---

### Key Metrics
- Tech vs Non-Tech orders (demand proxy)
- Tech vs Non-Tech revenue
- Average tech item price on Magist
- ENIAC average price benchmark (€540)

## Results — Demand, Revenue & Pricing

### 1️⃣ Demand (Orders)
- **Tech customers:** 18.5%
- **Non-Tech customers:** 81.5%

Tech orders account for a **minority of total demand** on Magist.  
Customer activity is primarily driven by **non-tech products**.

---

### 2️⃣ Revenue
- Tech products generate a **significantly smaller share of total revenue**
- Revenue concentration remains **outside the tech segment**

Despite platform growth, **tech does not act as a main revenue driver**.

---

### 3️⃣ Price Positioning
- **Average tech item price on Magist:** ≈ **€120**
- **ENIAC average product price:** **€540**

Magist operates in a **mid- to low-price range**, while ENIAC is clearly positioned in the **premium segment**.

---

## Business Interpretation
While Magist shows steady overall growth, the **tech segment lacks both scale and pricing power**.  
Current demand and revenue levels **do not support ENIAC’s premium positioning**.

---

## Conclusion (Slide-ready)
Magist’s tech segment:
- ❌ Represents a **minority of demand**
- ❌ Generates **limited revenue**
- ❌ Operates at prices **far below ENIAC’s premium level**

➡️ **Misalignment with ENIAC’s premium tech expansion strategy**
