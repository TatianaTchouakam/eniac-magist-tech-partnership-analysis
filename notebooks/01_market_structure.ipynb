# 01 — Market Structure (Magist)

## Goal
Assess whether Magist has a strong tech ecosystem by measuring the share of:
- Tech products
- Tech sellers
- Tech customers

## Expected Insight
Magist is a generalist marketplace rather than a tech-focused platform.

## Data & Definitions

### What is "Tech"?
Tech refers to categories such as computers, electronics, accessories and devices.

### Metrics
- % Tech products = tech_products / total_products
- % Tech sellers = tech_sellers / total_sellers
- % Tech customers = tech_customers / total_customers


## Analysis Plan
1. Create a mapping: product category → Tech vs Non-Tech
2. Compute the share of Tech products
3. Compute the share of Tech sellers
4. Compute the share of Tech customers
5. Visualize results (bar charts)
6. Write a business conclusion


## Results — Market Structure

Based on the analysis of the Magist marketplace:

- **Tech products represent 15.2%** of the total catalog (5,018 out of 32,951 products).
- **Tech sellers account for 20.0%** of all sellers (618 out of 3,095 sellers).
- **Only 18.5% of customers** purchase tech products (18,205 out of 98,666 customers).

## Business Insight

Magist is a generalist marketplace, where the tech segment represents a minority across products, sellers, and customers.


## Conclusion (to be confirmed by data)
If Tech represents only a small share across products, sellers and customers,
Magist cannot be considered a tech-focused marketplace.
